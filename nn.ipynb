```python
import pandas as pd

# Load the Excel file
df = pd.read_excel('nadil_category_expenses.xlsx')  # Replace 'your_file.xlsx' with your actual file path

# Set initial balance
initial_balance = 23394.64

# Calculate the Balance based on Payments, starting with the initial balance
df['Balance'] = initial_balance + df['Payments'].cumsum()

# Display the updated DataFrame
print(df[['Date', 'Description', 'Payments', 'Receipts', 'Balance']])

# Optionally, you can save the updated DataFrame back to Excel
df.to_excel('updated_file.xlsx', index=False)  # Save the updated file with a new name
```
